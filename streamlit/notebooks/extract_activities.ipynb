```python
import json
import re
import pandas as pd

from bs4 import BeautifulSoup

output_list = []
```

```python
full_file_name = "activities-updated.json"
with open(full_file_name, "r") as file:
    json_content = json.load(file)
df = pd.DataFrame(json_content)


def extract_page_content(steps):
    page_content = ""
    for step in steps:
        if isinstance(step, dict):
            text = step.get("text")
            if text:
                page_content += text + "\n"
            editor_state = step.get("editorState")
            if editor_state:
                page_content += editor_state + "\n"

    soup = BeautifulSoup(page_content)
    text = soup.get_text()
    text = text.strip()
    return re.sub(r"(\r\n.?)+", r"\r\n", text)


df["pageContent"] = df["steps"].apply(extract_page_content)
# df = df[["activityId", "activityName", "pageContent"]]

df.to_parquet("activities.parquet")
```

```python
import json
from pathlib import Path

files = Path("../app-streamlit/resources/images").glob("*.png")
files = list(files)

for file in files:
    if file.stem == "lwp-logo":
        continue
    name = file.stem
    meh = df[df["activityId"] == int(name)]
    print(meh.shape)
    if meh.shape[0] > 0:
        with open(f"../app-streamlit/resources/data/{name}.json", "wb") as f:
            f.write(json.dumps(meh["steps"].iloc[0]).encode("utf-8"))
```
